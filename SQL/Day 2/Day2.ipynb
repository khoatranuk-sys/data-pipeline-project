```python
from google.cloud import bigquery

client = bigquery.Client(project="bq-learning-510104")

def check(sql):
    """Chỉ ước lượng, không tốn dữ liệu."""
    dry = client.query(sql, job_config=bigquery.QueryJobConfig(dry_run=True))
    print(f"Sẽ quét: {dry.total_bytes_processed / 1024 / 1024:.2f} MB")

def run(sql, max_mb=100):
    """Chạy thật, có giới hạn an toàn."""
    config = bigquery.QueryJobConfig(maximum_bytes_billed=max_mb * 1024 * 1024)
    return client.query(sql, job_config=config).to_dataframe()
```

```python
sql = """
Create or replace table 'dim_users'

select
*
from `bq-learning-510104.dwh.dim_users`
limit 100
"""
```

```python
check(sql)
```

```python
df = run(sql)
df
```
